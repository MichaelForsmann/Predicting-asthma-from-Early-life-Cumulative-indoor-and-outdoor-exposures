```python
fig,ax=plt.subplots(3,3,figsize=(15,15))
sns.kdeplot(data_7['performance_test']["roc_test"].mean(axis=0),ax=ax[0,0],fill=True,alpha=0.5,label="7 years adjusted ROC")
sns.kdeplot(data_12['performance_test']["roc_test"].mean(axis=0),ax=ax[0,0],fill=True,alpha=0.5,label="12 years adjusted ROC")
sns.kdeplot(data_18['performance_test']["roc_test"].mean(axis=0),ax=ax[0,0],fill=True,alpha=0.5,label="18 years adjusted ROC")
ax[0,0].set_xlabel("Auroc")
ax[0,0].legend(loc=1)
sns.kdeplot(data_7['performance_test']["bss_test"].mean(axis=0),ax=ax[0,1],fill=True,alpha=0.5,label="7 years adjusted BSS")
sns.kdeplot(data_12['performance_test']["bss_test"].mean(axis=0),ax=ax[0,1],fill=True,alpha=0.5,label="12 years adjusted BSS")
sns.kdeplot(data_18['performance_test']["bss_test"].mean(axis=0),ax=ax[0,1],fill=True,alpha=0.5,label="18 years adjusted BSS")
ax[0,1].legend(loc=1)


ax[0,1].set_xlabel("bss")
sns.kdeplot(data_7['performance_test']["ici_test"].mean(axis=0),ax=ax[0,2],fill=True,alpha=0.5,label="7 years adjusted ICI")
sns.kdeplot(data_12['performance_test']["ici_test"].mean(axis=0),ax=ax[0,2],fill=True,alpha=0.5,label="12 years adjusted ICI")

sns.kdeplot(data_18['performance_test']["ici_test"].mean(axis=0),ax=ax[0,2],fill=True,alpha=0.5,label="18 years adjusted ICI")

ax[0,2].set_xlabel("ICI")
ax[0,2].legend(loc=1)


sns.kdeplot(data_7['performance_test']["roc_test"].mean(axis=0),ax=ax[1,0],fill=True,alpha=0.5,label="7 years ROC")
sns.kdeplot(data_12['performance_test']["roc_test"].mean(axis=0),ax=ax[1,0],fill=True,alpha=0.5,label="12 years ROC")
sns.kdeplot(data_18['performance_test']["roc_test"].mean(axis=0),ax=ax[1,0],fill=True,alpha=0.5,label="18 years ROC")
ax[1,0].set_xlabel("Auroc")
ax[1,0].legend(loc=1)
sns.kdeplot(data_7['performance_test']["bss_test"].mean(axis=0),ax=ax[1,1],fill=True,alpha=0.5,label="7 years BSS")
sns.kdeplot(data_12['performance_test']["bss_test"].mean(axis=0),ax=ax[1,1],fill=True,alpha=0.5,label="12 years BSS")
sns.kdeplot(data_18['performance_test']["bss_test"].mean(axis=0),ax=ax[1,1],fill=True,alpha=0.5,label="18 years BSS")
ax[1,1].legend(loc=1)


ax[1,1].set_xlabel("bss")
sns.kdeplot(data_7['performance_test']["ici_test"].mean(axis=0),ax=ax[1,2],fill=True,alpha=0.5,label="7 years ICI")
sns.kdeplot(data_12['performance_test']["ici_test"].mean(axis=0),ax=ax[1,2],fill=True,alpha=0.5,label="12 years ICI")

sns.kdeplot(data_18['performance_test']["ici_test"].mean(axis=0),ax=ax[1,2],fill=True,alpha=0.5,label="18 years ICI")

ax[1,2].set_xlabel("ICI")
ax[1,2].legend(loc=1)






sns.kdeplot(data_12['performance_test']["roc_test"].T,ax=ax[2,0],fill=True,alpha=0.5,label="12 years ROC")

ax[2,0].set_xlabel("Auroc")
ax[2,0].legend(loc=1)

sns.kdeplot(data_12['performance_test']["bss_test"].T,ax=ax[2,1],fill=True,alpha=0.5,label="12 years BSS")

ax[2,1].legend(loc=1)


ax[2,1].set_xlabel("bss")

sns.kdeplot(data_12['performance_test']["ici_test"].T,ax=ax[2,2],fill=True,alpha=0.5,label="12 years ICI")


ax[2,2].set_xlabel("ICI")
ax[2,2].legend(loc=1)
```
