```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr
import pickle
import arviz as az
```

```python


paths = [f"Models/asthma_7_year_repeat{j}_fold{i}.nc" for i in range(3) for j in range(2)]
trees = [az.from_netcdf(p) for p in paths]

# Stack the parameter posteriors across folds
combined_posterior_7 = xr.concat([t["posterior"].to_dataset() for t in trees], dim="chain",
)

list_parameters=["$NO_x$","$NO_2$","Formaldehyd","Acetone","Acetald",'constrution_year',"Black Carbon","$PM_{2.5}$","Street $NO_2$","Street $NO_x$","Street $PM_{2.5}$","Gestational age","Polygenetic risk score"]
combined_posterior_7 = combined_posterior_7.assign_coords(weight_dim_0=list_parameters)
combined_posterior_7["weight_OR"]=np.exp(combined_posterior_7["weight"])
asthma_7 = xr.DataTree.from_dict({"posterior": combined_posterior_7})
```

```python
weights = combined_posterior_12["weight_OR"].stack(sample=("chain", "draw")).values  # (D, S)

# 50% and 95% credible intervals
q025, q25, q50, q75, q975 = np.quantile(weights, [0.025, 0.25, 0.5, 0.75, 0.975], axis=1)




order = np.argsort(q50)
y = np.arange(len(list_parameters))
labels_sorted = [list_parameters[i] for i in order]
colors = ["C3" if q50[i] > 1 else "C0" for i in order]  # red for risk, blue for protective

fig, ax = plt.subplots(3,1,figsize=(7, 21))

for k, i in enumerate(order):
    ax[0].hlines(y[k], q025[i], q975[i], color=colors[k], lw=1)
    ax[0].hlines(y[k], q25[i], q75[i], color=colors[k], lw=4)
    ax[0].plot(q50[i], y[k], "o", color="white",
            markeredgecolor=colors[k], markersize=7, zorder=3)

ax[0].axvline(1.0, color="k", ls="--", lw=1)

ax[0].set_yticks(y); ax[0].set_yticklabels(labels_sorted)
ax[0].set_xlabel("Odds ratio per 1 SD")
ax[0].grid(axis="x", alpha=0.3)
plt.tight_layout()
```
