```python
import pandas as pd 
from tableone import TableOne
```

```python
df = pd.read_excel('../Data_preprocessing/all_time.xlsx', index_col=0)
```

```python
df["asthma_ever"]=df.loc[:,['asthma_0_7yrs', 'verificeret_12y', 'verificeret_18y']].sum(axis=1)>0
```

```python
df.loc[:,['asthma_0_7yrs', 'verificeret_12y', 'verificeret_18y']].isna().sum()
```

```python
df=df.rename(dict(zip(["mother's_education_1yr"],["Education"])),axis=1)
```

```python
columns=['scaled_nox', 'scaled_no2', 'scaled_formaldehyd', 'scaled_acetone',
       'scaled_acetald', 'scaled_bc', 'scaled_PM25', 'no2_street', 'nox_street',
       'pm25_street', 'bmi_6_yrs', 'gestational_age',
       'lrti_1y', 'genetic_risk_score_asthma', 'PCA1', 'PCA2','asthma_ever', "Education",'constrution_year', 'Area', 'floor_level',
       'powerplant', 'road', 'motorway', 'forest', 'coastline', 'airport',
       'farmland', 'industrial', 'carpet', 'cookerhood', 'passive', 'gastove',
       'fireplace']
categorical = ['asthma_ever']
continuous =  ['scaled_nox', 'scaled_no2', 'scaled_formaldehyd', 'scaled_acetone',
       'scaled_acetald', 'scaled_bc', 'scaled_PM25', 'no2_street', 'nox_street',
       'pm25_street','constrution_year', 'Area', 'floor_level',
       'powerplant', 'road', 'motorway', 'forest', 'coastline', 'airport',
       'farmland', 'industrial', 'carpet', 'cookerhood', 'passive', 'gastove',
       'fireplace']
groupby=  "Education"

nonnormal = continuous
```

```python
mytable = TableOne(df, columns, categorical, continuous, groupby,  nonnormal=nonnormal,pval=True, htest_name=True)
```

```python
mytable.to_latex('../Data_preprocessing/mytable_2.tex')
```

```python
mytable.to_html('../Data_preprocessing/mytable_2.html')
```
